```python
def normalize(x):
    if not isinstance(x, str):
        return None
    return (
        x.lower()
         .strip()
         .replace('*','')
         .replace(' *','')
         .replace('.','')
         .replace('(','')
         .replace(')','')
         .replace('-',' ')
    )
```

```python
monthly_data = (
    df_cleaned.groupby(['year', 'month'])['total_enroll']
    .sum()
    .reset_index()
)

monthly_data['date'] = pd.to_datetime(
    monthly_data[['year', 'month']].assign(day=1)
)

monthly_data = monthly_data.sort_values('date')
```

```python
monthly_data['moving_avg_3'] = (
    monthly_data['total_enroll']
    .rolling(window=3)
    .mean()
)
```

```python
plt.figure(figsize=(10,5))
plt.plot(monthly_data['date'], monthly_data['total_enroll'], label='Actual Enrolments')
plt.plot(monthly_data['date'], monthly_data['moving_avg_3'], label='3-Month Forecast')
plt.xlabel("Date")
plt.ylabel("Total Enrolments")
plt.title("Aadhaar Enrolment Trend and Forecast")
plt.legend()
plt.show()
```

```python
last_avg = monthly_data['moving_avg_3'].iloc[-1]

future_dates = pd.date_range(
    start=monthly_data['date'].iloc[-1] + pd.offsets.MonthBegin(1),
    periods=3,
    freq='MS'
)

future_forecast = pd.DataFrame({
    'date': future_dates,
    'forecast_enrolments': [last_avg] * 3
})
```

```python
plt.figure(figsize=(10,5))
plt.plot(monthly_data['date'], monthly_data['total_enroll'], label='Actual')
plt.plot(monthly_data['date'], monthly_data['moving_avg_3'], label='Historical Forecast')
plt.plot(future_forecast['date'], future_forecast['forecast_enrolments'],
         linestyle='dashed', label='Future Forecast')
plt.legend()
plt.title("Future Aadhaar Enrolment Forecast")
plt.show()
```
